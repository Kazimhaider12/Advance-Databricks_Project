```python
from pyspark.sql.functions import col, when, trim

# Target Schema Name
schema_name = "default"

# Saari 14 Tables ki List
bronze_tables = [
    "circuits", "constructors", "drivers", "races", "results",
    "status", "seasons", "qualifying", "driver_standings",
    "constructor_standings", "constructor_results", "pit_stops",
    "lap_times", "sprint_results"
]

cleaned_dataframes = {}

for table in bronze_tables:
    raw_table_name = f"self_databricks_account.{schema_name}.bronze_{table}"
    
    try:
        df_raw = spark.table(raw_table_name)
        df_clean = df_raw
        
        # 1. Clean String '\N' & Trim Spaces across all columns
        for c in df_raw.columns:
            df_clean = df_clean.withColumn(
                c, 
                when(trim(col(c)) == "\\N", None).otherwise(trim(col(c)))
            )
        
        # 2. Drop Exact Duplicate Rows
        df_clean = df_clean.dropDuplicates()
        
        # 3. Drop rows where Primary Key (e.g. driverId, raceId) is NULL
        pk_columns = [c for c in df_clean.columns if c.lower().endswith("id")]
        if pk_columns:
            df_clean = df_clean.dropna(subset=[pk_columns[0]])
            
        cleaned_dataframes[table] = df_clean
        print(f"Cleaned: bronze_{table} | Remaining Rows: {df_clean.count()}")
        
    except Exception as e:
        print(f"Error cleaning bronze_{table}: {e}")
```

```python
from pyspark.sql.functions import col, when, trim, concat_ws, to_date, coalesce, lit

schema_name = "default"

def clean_str(c):
    return when(trim(col(c)) == "\\N", None).otherwise(trim(col(c)))

df_drivers_raw = spark.table(f"self_databricks_account.{schema_name}.bronze_drivers")
before = df_drivers_raw.count()

df_drivers = (
    df_drivers_raw
    .dropDuplicates(["driverId"])
    .select(
        clean_str("driverId").cast("int").alias("driverId"),
        clean_str("driverRef").alias("driverRef"),
        
        # NULL values ko 0 se replace karne ke liye coalesce use kiya hai
        coalesce(clean_str("number").cast("int"), lit(0)).alias("number"),
        
        clean_str("code").alias("code"),
        clean_str("forename").alias("forename"),
        clean_str("surname").alias("surname"),
        concat_ws(" ", clean_str("forename"), clean_str("surname")).alias("driver_name"),
        to_date(clean_str("dob"), "yyyy-MM-dd").alias("dob"),
        clean_str("nationality").alias("nationality")
    )
    .dropna(subset=["driverId"])
)

```

```python
display(df_drivers)
```

```python
df_races_raw = spark.table(f"self_databricks_account.{schema_name}.bronze_races")
before = df_races_raw.count()

df_races = (
    df_races_raw
    .dropDuplicates(["raceId"])
    .select(
        clean_str("raceId").cast("int").alias("raceId"),
        clean_str("year").cast("int").alias("year"),
        clean_str("round").cast("int").alias("round"),
        clean_str("circuitId").cast("int").alias("circuitId"),
        clean_str("name").alias("name"),
        to_date(clean_str("date"), "yyyy-MM-dd").alias("date")
        # Unwanted date columns automated filter out ho chuke hain
    )
    .dropna(subset=["raceId", "circuitId"])
)
```

```python
display(df_races)
```

```python
df_results_raw = spark.table(f"self_databricks_account.{schema_name}.bronze_results")
before = df_results_raw.count()

df_results = (
    df_results_raw
    .dropDuplicates(["resultId"])
    .select(
        clean_str("resultId").cast("int").alias("resultId"),
        clean_str("raceId").cast("int").alias("raceId"),
        clean_str("driverId").cast("int").alias("driverId"),
        clean_str("constructorId").cast("int").alias("constructorId"),
        clean_str("number").cast("int").alias("number"),
        clean_str("grid").cast("int").alias("grid"),
        clean_str("position").cast("int").alias("position"),
        clean_str("points").cast("float").alias("points"),
        clean_str("laps").cast("int").alias("laps"),
        )
    .dropna(subset=["resultId", "raceId", "driverId", "constructorId"])
)
```

```python
df_circuits_raw = spark.table(f"self_databricks_account.{schema_name}.bronze_circuits")
before = df_circuits_raw.count()

df_circuits = (
    df_circuits_raw
    .dropDuplicates(["circuitId"])
    .select(
        clean_str("circuitId").cast("int").alias("circuitId"),
        clean_str("name").alias("name"),
        clean_str("location").alias("location"),
        clean_str("country").alias("country"),
        clean_str("lat").cast("float").alias("lat"),
        clean_str("lng").cast("float").alias("lng"),
        clean_str("alt").cast("int").alias("alt")
    )
    .dropna(subset=["circuitId"])
)
```

```python
df_lap_times_raw = spark.table(f"self_databricks_account.{schema_name}.bronze_lap_times")
before = df_lap_times_raw.count()

df_lap_times = (
    df_lap_times_raw
    .dropDuplicates()
    .select(
        clean_str("raceId").cast("int").alias("raceId"),
        clean_str("driverId").cast("int").alias("driverId"),
        clean_str("lap").cast("int").alias("lap"),
        clean_str("position").cast("int").alias("position"),
        clean_str("milliseconds").cast("int").alias("milliseconds")
    )
    .dropna(subset=["raceId", "driverId"])
)
```
