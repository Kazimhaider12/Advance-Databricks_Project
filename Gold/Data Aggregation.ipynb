```python
%sql
SELECT 
    result.driver_number AS Driver_Number,
    driver.driver_name AS Name,
    driver.nationality AS Nationality,
    constructor.team_name AS Brand_Name,
    circuit.location AS Location,
    circuit.country AS Country,
    laps.lap_number AS Lap_Number,
    laps.position AS Lap_Position, 
    result.laps_completed AS Laps,
    race.race_year AS Race_Year,
    race.race_round AS Race_Round,
    result.final_position AS Final_Position
FROM self_databricks_account.default.silver_results AS result
LEFT JOIN self_databricks_account.default.silver_drivers AS driver
    ON result.driver_id = driver.driver_id
LEFT JOIN self_databricks_account.default.silver_races AS race
    ON result.race_id = race.race_id
LEFT JOIN self_databricks_account.default.silver_constructors AS constructor
    ON result.constructor_id = constructor.constructor_id
LEFT JOIN self_databricks_account.default.silver_lap_times AS laps
    ON result.race_id = laps.race_id 
   AND result.driver_id = laps.driver_id
left join self_databricks_account.default.silver_circuits AS circuit
    ON race.circuit_id = circuit.circuit_id
ORDER BY Final_Position ASC NULLS LAST;
```
